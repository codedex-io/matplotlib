```python
import matplotlib.pyplot as plt
import numpy as np

import pandas as pd

df = pd.read_csv('./data/gapminder.csv')

plt.figure()
plt.scatter(df['lifeExp'], df['gdpPercap'])
plt.title('Life Expectancy vs GDP per capita')
plt.xlabel('Life Expectancy')
plt.ylabel('GDP per capita')
plt.show()
```
