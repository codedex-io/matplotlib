```python
import matplotlib.pyplot as plt
import numpy as np 

import pandas as pd

df = pd.DataFrame({
    'aquarium': [
        'MonterreyBay',
        'GeorgiaAquarium',
        'SheddAquarium',
        'NationalAquarium'
    ],
    'Shark': [5, 4, 4, 2],
    'Dolphin': [8, 12, 10, 5],
    'Octopus': [7, 20, 8, 4],
    'Sea Turtle': [12, 5, 6, 3],
    'Jellyfish': [30, 25, 20, 15]
})

plt.figure()
plt.bar(df['aquarium'], df['Shark'])

plt.title('Sharks in Aquariums')
plt.xlabel('Aquarium')
plt.ylabel('Number of Sharks')
plt.show()
```
