```python
import matplotlib.pyplot as plt
import numpy as np

import pandas as pd

df = pd.DataFrame({
  'toppings': ['Pepperoni', 'Cheese', 'Mushroom', 'Pineapple', 'Prosciutto', 'Artichoke'],
  'orders': [40, 25, 20, 15, 4, 5]
})

plt.figure()
plt.pie(df['orders'], labels=df['toppings'])
plt.title('Favorite Pizza Toppings')
plt.xlabel('Toppings')
plt.show()
```
