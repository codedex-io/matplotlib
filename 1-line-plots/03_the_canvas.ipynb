```python
import matplotlib.pyplot as plt
import numpy as np
```

## Solution

Visualize your energy levels throughout the day.

```python
hours = np.array([8, 10, 12, 14, 16, 18, 20])
energy = np.array([6, 8, 7, 4, 5, 7, 3])

plt.figure()
plt.plot(hours, energy)
plt.show()
```
