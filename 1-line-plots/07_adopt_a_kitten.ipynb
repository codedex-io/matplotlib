```python
import matplotlib.pyplot as plt
import numpy as np

import pandas as pd

df = pd.DataFrame({
  'month': ['Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'],
  'adoptions': [45000, 52000, 48000, 43000, 38000, 41000]
})

plt.figure()
plt.plot(df['month'], df['adoptions'], color='red')
plt.title('Adoption Rate of Kittens')
plt.xlabel('Month')
plt.ylabel('Adoptions')
plt.show()
```
