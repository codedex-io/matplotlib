Exercise link: https://www.codedex.io/matplotlib/02-screen-time

```python
import matplotlib.pyplot as plt
import numpy as np

days = np.array(['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'])
screen_time = np.array([120, 100, 110, 130, 140, 150, 160])

plt.figure()
plt.plot(days, screen_time)
plt.show()
```
