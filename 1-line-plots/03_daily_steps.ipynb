# 03. Daily Steps 🚶‍♂️
Exercise link: https://www.codedex.io/matplotlib/03-daily-steps

```python
import matplotlib.pyplot as plt
import numpy as np

days = np.array(['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'])
steps = np.array([3000, 4000, 2500, 5000, 4500, 3500, 6000])

plt.figure()
plt.plot(days, steps)
plt.title('Daily Steps')
plt.xlabel('Day')
plt.ylabel('Steps')
plt.show()
```
