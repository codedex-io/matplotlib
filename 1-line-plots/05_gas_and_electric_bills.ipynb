Exercise link: https://www.codedex.io/matplotlib/04-gas-and-electric-bills

```python
import matplotlib.pyplot as plt
import numpy as np

months = np.array(['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'])
gas_bills = np.array([45, 48, 52, 55, 58, 60, 62, 65, 68, 70, 72, 75])
electric_bills = np.array([120, 115, 125, 130, 135, 140, 145, 150, 155, 160, 165, 170])

plt.figure()
plt.plot(months, gas_bills, label='Gas Bills')
plt.plot(months, electric_bills, label='Electric Bills')
plt.title('Gas & Electric Bills')
plt.xlabel('Month')
plt.ylabel('Bills')
plt.legend()
plt.show()
```
