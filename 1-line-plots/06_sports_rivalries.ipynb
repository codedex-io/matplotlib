```python
import matplotlib.pyplot as plt
import numpy as np

games = np.array(['1', '2', '3', '4', '5'])

bulls_points = np.array([102, 105, 110, 108, 106])
celtics_points = np.array([98, 100, 102, 104, 101])
warriors_points = np.array([110, 115, 108, 112, 111])
lakers_points = np.array([108, 112, 109, 114, 113])

plt.figure()
plt.plot(games, bulls_points, label='Bulls', color='red', linestyle='--')
plt.plot(games, celtics_points, label='Celtics', color='blue', linestyle='--')
plt.plot(games, warriors_points, label='Warriors', color='green', linestyle='--')
plt.plot(games, lakers_points, label='Lakers', color='yellow', linestyle='--')

plt.title('Bulls vs Celtics vs Warriors vs Lakers')
plt.xlabel('Game')
plt.ylabel('Points')
plt.legend()
plt.show()
```
