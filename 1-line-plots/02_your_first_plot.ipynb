# 02. Your First Plot

Learn how to import Matplotlib and create your very first line plot.

```python
import matplotlib.pyplot as plt
import numpy as np
```

## Solution

Create a line plot of your screen time for the past week.

```python
days = np.array(['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'])
screen_time = np.array([120, 180, 150, 130, 160, 170, 140])

plt.plot(days, screen_time)
plt.show()
```
